**Loops and List Comprehensions**

* For loop: ONCE for every item in the list
* Print function : built in vertical output each line
* end: keyword argument, parameter, socket
* /n : New line
* end = ' '

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet)
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet, end = '---')
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet, end = '>>>')
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet, end = '\n\n')
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet, end = '\n\n\n\n\n')
```

**While Loop 'i'**

**i += 2: ADD & ASSIGN**
* increase i by 2
* original : i = i + 2
* shortcut : i += 2

**Signs**
* i += 1 (Addition: Increase by 1)
* i -= 1 (Subtraction: Decrease by 1)
* i *= 2 (Multiplication: Double it)
* i /= 2 (Division: Halve it)



```python
i = 0
while i < 10:
    print(i)
    i += 2  # The "Jump"
```

```python
i = 0
while i < 5:
    print(i, end = '; ')
    i += 1.5  # The "Jump"
```

**For Loop Shortcut Microwave: List Comprehensions**

* n**2: n^2: n powered to the 2
* n**3: n^3: n powered to the 3

```python
numbers = [1, 2, 3]
doubled = []
for n in numbers:
    doubled.append(n * 2)
    print(doubled)
print(doubled)
```

**What's inside?**

* Interpreter environment
* variable name as the very last line
* Show preview of result
* Courtesy

```python
doubled = [n * 2 for n in numbers]
doubled
```

```python
squares = [n**2 for n in range(10)]
squares
```

```python
squares = []
for n in range(10):
    squares.append(n**2)
squares
```

**LIST: 'if' condition**

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
short_planets = [planet for planet in planets if len(planet) < 6]
short_planets
```

**LIST: 'if condition and loop variable**

```python
# str.upper() returns an all-caps version of a string
loud_short_planets = [planet.upper() + '!' for planet in planets if len(planet) < 6]
loud_short_planets
```
