**Loops and List Comprehensions**

* For loop: ONCE for every item in the list
* Print function : built in vertical output EACH line
* end: keyword argument, parameter, socket
* /n : New line
* end = ' '

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet)
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet, end = '---')
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet, end = '>>>')
```

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
for planet in planets:
    print(planet, end = '\n\n')
```

**For Loop Shortcut Microwave: List Comprehensions**

* n**2: n^2: n powered to the 2
* n**3: n^3: n powered to the 3

```python
numbers = [1, 2, 3]
doubled = []
for n in numbers:
    doubled.append(n * 2)
    print(doubled)
doubled
```

**What's inside?**

**LIST: 'if' condition**

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
short_planets = [planet for planet in planets if len(planet) < 6]
short_planets
```

**LIST: 'if condition and loop variable**

```python
# str.upper() returns an all-caps version of a string
loud_short_planets = [planet.upper() + '!' for planet in planets if len(planet) < 6]
loud_short_planets
```

**One line LIST Microwave**

* min, max, sum

```python
def count_negatives(nums):
    """Return the number of negative numbers in the given list.

    >>> count_negatives([5, -1, -2, 0, 3])
    2
    """
    n_negative = 0
    for num in nums:
        if num < 0:
            n_negative = n_negative + 1
    return n_negative

```

```python
def count_negatives(nums):
    return len([num for num in nums if num < 0])
```

```python
def count_negatives(nums):
    return sum([num < 0 for num in nums])
```
